#Data Reading

```python
df = spark.read.format("parquet")\
    .option("inferSchema", True)\
    .load("abfss://carsproject@pravdatalake.dfs.core.windows.net/bronze")
```

```python
df.display()
```

## Data Transformation 

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
df = df.withColumn('model_category', split(col('Model_ID'),'-')[0])
```

```python
df.display()
```

```python
df.withColumn('Units_Sold', col('Units_Sold').cast(StringType())).printSchema()
```

```python
df = df.withColumn('RevPerUnit',col('Revenue')/col('Units_Sold'))
df.display()
```

##AD-HOC

```python
display(df.groupBy('Year','BranchName').agg(sum(col('Units_Sold')).alias('Total_Units')).sort('Year','Total_Units', ascending=[1,0]))
```

##Data Writing

```python
df.write.format('parquet')\
    .mode('overwrite')\
    .option('path', 'abfss://carsproject@pravdatalake.dfs.core.windows.net/silver')\
        .save()
```
